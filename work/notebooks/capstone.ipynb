# Capstone — Refresh Opportunity Scoring

**Scope:** bundled anonymized starter slice (30,000 content items across 32 pseudonymized clients).

## Abstract

This project asks which visible content items an editor should review first. It uses the bundled anonymized FlyRank starter slice of 30,000 pseudonymized content items and a grouped client holdout. A transparent visibility-and-freshness baseline is compared with logistic regression and random forest models using a current-window decline proxy and Precision@K. The resulting metrics quantify out-of-sample ranking quality on the held-out clients, while the output is a public-safe, reason-coded review queue. This is directional decision support, not a prediction of Google's algorithm or evidence that editing causes recovery.

## 1. Question

For editors deciding what to review first, rank visible pages by an observed decline-risk proxy.

## 2. Data

The analysis uses `content_refresh_anonymized.csv`: one row per pseudonymized content item, 30,000 rows and 32 clients. IDs are split-only context; label-derived trend fields, provider/model fields, and any product flags are excluded.

## 3. Methodology

The target is `trend_direction == down`; `trend_direction` and `trend_pct` are never features. Numeric values use median imputation; categoricals use most-frequent imputation plus one-hot encoding. The split holds out clients, and baseline/model metrics use the same test rows.

## 5. Limitations
