# ML-03 — Frame Your Lane as an ML Task

**Scope:** bundled anonymized starter slice (30,000 content items across 32 pseudonymized clients).

## 1. Task type

This is binary classification used for **ranking**: estimate the probability that a page has the observed decline proxy, then rank the review queue.

## 2. Target / proxy

`is_declining_label = 1` when `trend_direction == 'down'`. This is an observed current-window proxy, not a future label; it is deliberately kept out of the features.

## 3. Success metric

Primary: Precision@50 on a client holdout, because a review team can act on a finite top-K list. The positive base rate is shown alongside every metric.

## 4. Unit of analysis

One row is one pseudonymized content item with trailing-90-day measurements.

## 5. Why ML

Freshness, visibility, position, CTR, engagement, and content context interact in ways a single threshold cannot express.
